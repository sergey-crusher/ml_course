# Домашнее задание №1: Задачи машинного обучения

### Вопросы и задания:
1. Какие выделяют типы машинного обучения? В чём их отличия?
2. Подумайте, к какому виду машинного обучения относится обучение ИИ игре в шахматы?

### Процесс выполнения:

1. Машинное обучение бывает с учителем и без учителя. С учителем подразумевает использование некоего эталона, к которому в процессе нужно максимально приблизиться. Собственно, когда результат будет в пределах статистической погрешности, обучение считается состоявшимся. В отличие от этого метода, при обучении без учителя, верный ответ заранее неизвестен. Происходит ряд некоторых допущений, общая задача разбивается на более мелкие и на основе практического смысла и здравости рассуждений происходит поиск наиболее верного решения.

2. Игра в шахматы – это сложный вычислительный процесс, предусматривающий многомерную технологию (количество вариантов продолжения партии огромно) и точного ответа быть не может. Поэтому, происходит деление на более мелкие кластеры. Таким образом мы имеем дело с обучением без учителя. Это сильно упрощённая теория, на самом деле обучение игре в шахматы включает в себя другие подклассы ML.
