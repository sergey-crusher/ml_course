#### Считываем данные из excel файла

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel("./data/L2.xls")
```

```python
data = df["Мужчины, млн человек"] # считываем колонку с этим названием
```

```python
data.head() # проверяем что данные считались
```

#### Построение/обучение модели
Возьмём среднее как самую простую модель

```python
model = float(data.mean())
print('{:.2f}'.format(model))
```

#### Вычисление точности модели
Рассчитываем ошибку по тестовым данным

```python
err = np.abs(data - model).mean()
print('{0:.3}%'.format(float(100*err/model)))
```

```python
# 2021 - 67,85 млн.
result = 67.85
print('{0:.3}%'.format(100 * (result - model) / result))
```
