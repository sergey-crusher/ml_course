### Процесс выполнения:

1. Обучающая выборка (train) — часть данных, на которой модель обучается: по ней подбираются параметры и строится зависимость.
Проверочная выборка (test) — часть данных, которую модель не видела при обучении; на ней оценивают, насколько точно модель работает на новых, ранее не встречавшихся данных.

2. Решаем задачу как во втором уроке, но разбиваем данные на обучающую и проверочную выборки.

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split

df = pd.read_excel("./data/L2.xls")
train, test = train_test_split(df, test_size=0.2)
data_train = train["Мужчины, млн человек"]
data_test = test["Мужчины, млн человек"]
model = float(data_test.mean())
err = np.abs(data_train - model).mean()
print('{0:.3}%'.format(float(100*err/model)))
```
