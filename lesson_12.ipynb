# Домашнее задание №12: Метод наименьших квадратов

### 1. Суть метода максимального правдоподобия

Суть метода: составить по специальной формуле функцию правдоподобия $L$, и найти оценку параметра $\theta$ из условия максимизации функции правдоподобия (ФП) на определенной выборке $\{x_i\}$. Иногда ФП заменяют на логарифмическую функцию правдоподобия $l = \ln L$ (ЛФП), что облегчает расчеты (вычисление производных).

Оценки, полученные данным методом, будут состоятельными, асимптотически эффективными и асимптотически нормальными. Несмещенность оценок надо проверять (этот метод не гарантирует).

### 2. Пример: оценка перекоса монеты (Метод максимального правдоподобия)

**Задача:** Мы подбросили монету 10 раз. Выпало 7 орлов и 3 решки. Какова реальная вероятность выпадения орла ($\theta$)?

**Логика ММП:** Мы должны выбрать такое значение $\theta$, при котором вероятность получить *именно такую* выборку (7 орлов, 3 решки) была бы максимальной.

1. **Функция правдоподобия:** Вероятность каждого орла равна $\theta$, каждой решки — $(1-\theta)$. Так как броски независимы, перемножаем их:
$$L(\theta) = \theta^7 \cdot (1-\theta)^3$$

2. **Поиск максимума:** Чтобы найти пик этой функции (не подбирая числа вручную), берем логарифм (для упрощения) и приравниваем производную к нулю:
$$l(\theta) = 7\ln\theta + 3\ln(1-\theta)$$
$$\frac{7}{\theta} - \frac{3}{1-\theta} = 0 \;\Rightarrow\; \hat{\theta} = 0.7$$

**Вывод:** Самым «правдоподобным» значением оказалась наблюдаемая нами частота ($\frac{7}{10}$). Если бы мы предположили, что монета честная ($\theta=0.5$), вероятность увидеть наши данные была бы в 2 с лишним раза ниже, чем при $\theta=0.7$.

### 3. Метод наименьших квадратов

МНК — это математический метод, применяемый для решения различных задач, основанный на минимизации суммы квадратов отклонений некоторых функций от искомых переменных.
