### Вопросы и задания:
1. Перечислите метрики, применяемые при регрессионном анализе.
2. Объясните как вы понимаете расстояние Чебышева?
3. Расстояние Минковского.

### Процесс выполнения:

1. Евклидово расстояние, либо расстояние Манхэттена.

    Евклидово расстояние - это кратчайший путь между двумя точками в евклидовом пространстве. Формула Евклидового расстояния:

    $d_{12} = \sqrt{(x_1 - x_2)^2 + (y_1 - y_2)^2}$

    В случае Манхэттеновского расстояния - тут, как следует из названия, при движении от одного перекрестка к другому перекрестку в районе Манхэттена расстояние в автомобиле, очевидно, не является расстоянием по прямой между двумя точками. Это фактическое расстояние проезда и есть «манхэттенское расстояние». Расстояние Манхэттена также называют «расстоянием городского квартала» (City Block Distance).

    Манхэттенское расстояние между двумя точками a (x1, y1) и b (x2, y2) на двумерной плоскости:
   $d_{12} = \left| x_1 - x_2 \right| + \left| y_1 - y_2 \right|$

    Манхэттен (ладья по клеткам, без диагоналей): 7 шагов

    ```
    . . . . Б
    . . . → ↑
    . . → ↑ .
    А → ↑ . .
    ```


2. В шахматах король может ходить прямо, горизонтально или диагонально, поэтому король может перейти на любую из 8 соседних клеток на один шаг. Сколько шагов нужно королю, чтобы пройти от сетки (x1, y1) до сетки (x2, y2)? Это расстояние называется расстоянием Чебышева.

    Расстояние Чебышева между двумя точками a (x1, y1) и b (x2, y2) на двумерной плоскости:
    $d_{12} = max(\left| x_1 - x_2 \right|,\left| y_1 - y_2 \right|)$

    max(4, 3) = 4 шага

    ```
    . . . → Б
    . . ↗ . .
    . ↗ . . .
    А . . . .
    ```

3. Расстояние Минковского (метрика Минковского) — параметрическая метрика на евклидовом пространстве, которую можно рассматривать как обобщение евклидова расстояния и расстояния городских кварталов:

    $p(x,y) = \left( \sum_{i=1}^n |x_i-y_i|^p \right)^{\frac{1}{p}}$
