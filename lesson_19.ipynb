# Домашнее задание №19: Ансамбли моделей машинного обучения. Ансамбль стекинга. (Теория)

### 1. Определение - что подразумевают под ансамблем модели машинного обучения?

Ансамблем называется алгоритм, который состоит из нескольких алгоритмов машинного обучения, а процесс построения ансамбля называется ансамблированием.

### 2. Придумайте пример (общий) для ансамбля.

$a(x) = \frac{1}{n} \left( b_1(x) + \ldots + b_n(x) \right)$

### 3. Какие ансамбли являются наиболее распространенными?

- **Бэггинг (Bagging, Bootstrap Aggregating)** - Модели обучаются параллельно, каждая на своей бутстрап-выборке (случайная подвыборка с повторениями), ответы усредняются.
- **Бустинг (Boosting)** - Модели обучаются последовательно: каждая следующая фокусируется на ошибках предыдущих (подгоняется под остатки / увеличивает вес сложных объектов). Примеры: AdaBoost, Gradient Boosting, XGBoost, LightGBM, CatBoost. Борется со смещением (bias): последовательные добавки делают модель всё точнее. Сейчас это рабочий стандарт для табличных данных.
- **Стекинг (Stacking)** - Ответы базовых моделей (базового уровня) подаются как признаки в мета-модель верхнего уровня, которая учится, кому верить.

### 4. Что подразумевается под мажоритарным, усреднённым и взвешенным голосованием при выборе в ансамбль моделей?

Мажоритарное голосование предполагает вычисление наиболее значимого из  показателей выбранных моделей, тогда как усреднённое голосование учитывает среднеарифметическое значение показателей. Взвешенное голосование учитывает коэффициент (вес) каждого из параметров.
