# Домашнее задание №3: Потоки данных в машинном обучении

### Вопросы и задания:
1. Что такое парсинг данных и его значение в ML?
2. В чём сильные стороны лямбда архитектуры и её сложность для реализации?

### Процесс выполнения:

1. Парсинг – это, в общем понимании, разбиение крупного объекта информации на более мелкие составные части для его углублённого анализа. Парсинг данных – это то же самое, относящееся к большому объёму данных. Парсинг данных – это основа машинного обучения. Он позволяет выделить отдельные элементы из огромного числа общих и проанализировать каждый из них, составить классификацию и по общим признакам уже строить модель обучения.

2. Лямбда-архитектура — это паттерн проектирования систем обработки больших данных, созданный Натаном Марцем.

В ML она используется, когда нужно одновременно иметь:
- точные результаты на всех исторических данных (batch layer)
- свежие предсказания в реальном времени (speed layer)

**Batch Layer** (пакетный слой)
- Хранит все исходные данные (immutable — неизменяемые)
- Периодически (раз в сутки/неделю) запускает тяжёлые расчёты
- обучение моделей на полных исторических данных

Результат: готовые к выдаче агрегаты и веса моделей

Плюсы: высокая точность, устойчивость к ошибкам

Минусы: задержка в часах/днях

**Speed Layer** (скоростной слой)
- Работает только с новыми данными, пришедшими после последнего батча
- Лёгкие модели
- Обновляется постоянно (секунды/минуты)

Результат: real-time — свежие, но менее точные результаты

Плюсы: низкая задержка

Минусы: меньше точность, сложнее поддержка

**Serving Layer** (обслуживающий слой)

Индексирует результаты из обоих слоёв

При запросе объединяет: ответ = batch + speed

```
            Новые данные
        │                 │
        ▼                 ▼
  ┌───────────┐     ┌───────────┐
  │  Master   │     │   Speed   │
  │  Dataset  │     │   Layer   │
  └───────────┘     └─────┬─────┘
        │                 │
        ▼                 │
  ┌───────────┐           │
  │   Batch   │           │
  │   Layer   │           │
  └─────┬─────┘           │
        │                 │
        ▼                 ▼
  ┌─────────────────────────────┐
  │         Serving Layer       │
  └─────────────────────────────┘
```

Плюсы:

✅ Устойчивость: если сломался speed layer, работают batch-данные

✅ Точность: batch даёт корректные результаты

✅ Актуальность: speed даёт свежие данные

Минусы:

❌ Два пайплайна — пишешь и поддерживаешь один и тот же код дважды (один для batch, другой для speed)

❌ Синхронизация — нужно согласовывать результаты двух слоёв
