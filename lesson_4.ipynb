```python
%matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.pyplot import rcParams
rcParams['figure.figsize'] = 16, 8
```

```python
data = pd.read_csv('./data/SouthGermanCredit.asc', sep=r"\s+")
print(data.head())
```

```python
data.plot.scatter(x="laufzeit", y="hoehe", alpha=0.2)
plt.xlabel("Срок")
plt.ylabel("Сумма")
plt.show()
```

### Построение простой модели
Среднее арифметическое

```python
model = float(data["hoehe"].mean())
print("{0:.2f}".format(model))
```

### Вычисление точности модели
Ошибка по данным
